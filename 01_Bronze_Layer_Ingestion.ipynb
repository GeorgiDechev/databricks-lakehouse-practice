```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

# 1. Дефинираме схема за нашите сурови данни (Bronze Layer)
schema = StructType([
    StructField("transaction_id", IntegerType(), True),
    StructField("account_id", StringType(), True),
    StructField("amount", DoubleType(), True),
    StructField("status", StringType(), True)
])

# 2. Създаваме примерни данни (симулираме екстракт от core banking система)
data = [
    (1, "ACC_100", 150.50, "COMPLETED"),
    (2, "ACC_200", 500.00, "PENDING"),
    (3, "ACC_100", 12.00, "COMPLETED"),
    (4, "ACC_300", 999.99, "FAILED"),
    (5, "ACC_200", 45.00, "COMPLETED")
]

# 3. Създаваме Spark DataFrame (в Databricks обектът 'spark' е винаги зареден по подразбиране)
df_bronze = spark.createDataFrame(data, schema)

# 4. Визуализираме данните в табличен вид
display(df_bronze)
```
